```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM ,

dtype = torch.bfloat16
```

```python
save_path = "/content/drive/MyDrive/qwen_model"
tokenizer = AutoTokenizer.from_pretrained(save_path)

model = AutoModelForCausalLM.from_pretrained(
    save_path,
    torch_dtype=dtype,
    device_map="auto"
)

print("Reloaded successfully!")
```

# 1. Understanding inference
Inference is the process of using a trained model to make predictions or generate outputs.

For an LLM, the basic process is:
1. Convert the input text into token IDs.
2. Pass token IDs through the model.
3. Predict the next token.
4. Add that token to the input.
5. Repeat until a stopping condition is reached.
6. Decode generated token IDs back into text.
Important: LLMs generate text one token at a time, not one complete sentence at once.

```python
prompt = "Nvdia is the leading"

inputs = tokenizer(prompt, return_tensors="pt")
input_device = model.get_input_embeddings().weight.device
inputs = inputs.to(input_device)

with torch.inference_mode():
  outputs = model.generate(
      **inputs,
      max_new_tokens=50
  )

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
